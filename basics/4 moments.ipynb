# Moments
A Moment is a custom object within the FMA Library. A Moment represents a single frame within the match and contains details such as the ball location and location/velocity of every player on the pitch at that frame.

**Import FMA Library**

```python
from FootballMatchAnalysis.objects.match import Match
```

**Load Event and Tracking Data**

```python
DATADIR = './../data'
game_id = 1
```

```python
match = Match(DATADIR, game_id)
```

**Moment Object**

A Moment, as mentioned, represents a given frame during the match. Moment objects can be generated by passing a frame number to `Match.get_moment()`.

```python
frame = 100
moment = match.get_moment(frame)
```

**Visualizing a Moment**

```python
moment.plot_moment()
```

**Ball and Player Locations from a Moment**


```python
# Get ball
ball = moment.ball

# Print ball location
print(f"Ball is at ({ball.x}, {ball.y})")

# Get all players
players = match.players()

# Find a player and get current physical data
player = players[10]

print(f"Player {player.name} is at ({player.x}, {player.y})")
print(f"Player {player.name} has a velocity of ({player.vx}, {player.vy})")
```

**Get Team and Player in Possession**

```python
on_ball = moment.possession()

print(f"Player {player.name} ({player.team} Team) is on ball")
```
