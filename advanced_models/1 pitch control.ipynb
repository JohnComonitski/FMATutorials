# Pitch Control
The FMA Library builds on the great work by [Friends of Tracking](https://www.youtube.com/@friendsoftracking755). One of their greatest contributions is their pitch control diagrams. You can watch how they built their pitch control model in this [video](https://www.youtube.com/watch?v=5X1cSehLg6s). The FMA library builds on its work by making it simple to calculate pitch control for any given moment of a match.

**Import FMA Library**

```python
from FootballMatchAnalysis.objects.match import Match
```

**Load Event and Tracking Data**

```python
DATADIR = './../data'
game_id = 1
```

```python
match = Match(DATADIR, game_id)
```

**Generate Pitch Control for a Moment**

The Moment method `generate_pitch_control` will call a modified method within the Metrica library to accept a Moment object to generate pitch control. The pitch control data will then be stored in the Moment object for future use.

```python
# Get a Moment
moment = match.get_moment(100)

# Generate Pitch Control
moment.generate_pitch_control()
```

**Plot Pitch Control**

Using the results of `generate_pitch_control`, we will now plot the pitch control. Note, you do not need to call `generate_pitch_control` before you generate a pitch control plot. If it has not been called yet, it will be called automatically.

```python
# Plot a Moment
plot = moment.plot_moment()

# Overlay a Pitch Control Voronoi Diagram
plot = moment.plot_pitch_control(plot)
```
