# Packing
Packing in football quantifies the number of opposing players bypassed by a pass or dribble. First popularized by IMPECT and popularize during the 2016 Euros, packing allows us to measures a player's ability to play or receive line breaking passes.

**Events Library**

The FMA library includes a series of libraries that you can utilize to support your analysis. We will start by importing the `events` library.

```python
from FootballMatchAnalysis.analysis.events import *
```

**Find Players Packed By a Pass**

```python
# Get all passes
passes = match.get_events("PASS")

# Get a pass
p = passes.iloc[27]

# Number of players packed
packed = players_packed(match, p)
print(f"Players packed: {packed}")

# Players who were packed
players = get_players_packed(match, p)
for player in players:
    print(f"Player {player.name} was packed")
```

**Annotating a Packing Situation**

```python
# Get a pass
p = match.events.iloc[99]

# Look at drame
frame = p["Start Frame"]
moment = match.get_moment(frame)

# Plot moment
plot = moment.plot_moment()

# Get players packed
# The optional parameter `relevance_distance` helps determine how far away can be to still be counted as 'packed'
players = get_players_packed(match, p, relevance_distance=15)

# Mark players who were packed
for packed_player in players:
    #Find player in moment
    player = None
    for p in moment.home_team():
        if p.name == packed_player.name:
            player = p

    if player is not None:
        plot.draw_circle(player.x, player.y, r=2, c='b')
plot
```
