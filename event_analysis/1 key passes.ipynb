# Finding and Plotting Key Pass
Finding key passes, plotting the sequence of passes that lead to the key and ways to analyze the key passes. A key pass is the final pass that leads directly to a teammate’s shot. The shot does not have to go in to count as a key pass.

**Import FMA Library**

```python
from FootballMatchAnalysis.objects.match import Match
from FootballMatchAnalysis.objects.plot import Plot
```

**Load Event and Tracking Data**

```python
DATADIR = './../data'
game_id = 1
```

```python
match = Match(DATADIR, game_id)
```

**Get All Passes**

Let's start by getting and visualizing all the passes from the Home team. You will notice there are many passes. Our goal is to cut through the noise and find the key passes.

```python
plot = Plot()

# Get all passes
passes = match.get_events("PASS")

# Filter for only passes by home team
passes = passes[passes["Team"] == "Home"]

# Iterate over passes and plot them 
for i, p in passes.iterrows():
    plot.draw_event(p, c="#1F77B4", alpha=1)
plot
```

**Events Library**

The FMA library includes a series of libraries you can support to assist you in your analysis. We will start by importing the `events` library.

```python
from FootballMatchAnalysis.analysis.events import *
```

**Is Key Passs**

imported from the `events` library, `is_key_pass` will take a match object and an event as parameters and return `True` if the event is a key pass, otherwise it will return `False`

```python
p = passes.iloc[2]

plot = Plot()
plot.draw_event(p)

if is_key_pass(match, p):
    print("Is a Key Pass")
else:
    print("Is Not a Key Pass")
plot
```

```python
p = match.events.iloc[660]

plot = Plot()
plot.draw_event(p)

if is_key_pass(match, p):
    print("Is a Key Pass")
else:
    print("Is Not a Key Pass")
plot
```
