**Load Event and Tracking Data**

```python
match = Match(DATADIR, game_id)
```

**Events Library**

The FMA library includes a series of libraries that you can utilize to support your analysis. We will start by importing the `events` library.

```python
from FootballMatchAnalysis.analysis.events import *
```

**Is One-Two**

Imported from the `events` library, `is_one_two` takes a match object and an event as parameters and returns `True` if the event is a key pass; otherwise, it returns `False`. We will be using it as a filter to identify passes that are part of a One-Two sequence.`

```python
# Passes from Home Team
passes = match.get_events("PASS")
passes = passes[passes["Team"] == "Home"]

# Filter for Starts of One-Two Passes
one_twos = passes[passes.apply(
    lambda p: is_one_two(match, p),
    axis=1
)]
```

**Plotting a One-Two**

```python
# Set up plot
plot = Plot()
plot.write("One-Two's Made", -52, 32, justifification="left")
plot.write("Opened", -52, 30, c='r', justifification="left")
plot.write("-", -44, 30, justifification="left")
plot.write("Closed", -43, 30, c='b', justifification="left")

# Get a One-Two
one_two = one_twos.iloc[20]

# Get the following pass
next_p = next_pass(match, one_two)

# Find the path of the player that started the One-Two
from_player = match.get_player(one_two["From"])
from_player_path = match.path(from_player, one_two["Start Frame"], next_p["End Frame"])

# Plot One-Two
plot.draw_event(one_two, c='r')
plot.draw_event(next_p, c='b')
plot.draw_path(from_player_path, alpha=.15)
plot
```

**Plot All One-Twos**

```python
# Set up plot
plot = Plot()
plot.write("One-Two's Made", -52, 32, justifification="left")
plot.write("Opened", -52, 30, c='r', justifification="left")
plot.write("-", -44, 30, justifification="left")
plot.write("Closed", -43, 30, c='b', justifification="left")

# Iterate over each One-Two
for i, p in one_twos.iterrows():
    # Get the following pass
    next_p = next_pass(match, p)

    # Find the path of the player that started the One-Two
    from_player = match.get_player(p["From"])
    from_player_path = match.path(from_player, p["Start Frame"], next_p["End Frame"])

    # Add One-Two to Plot
    plot.draw_event(p, c='r')
    plot.draw_event(next_p, c='b')
    plot.draw_path(from_player_path, alpha=.15)
plot
```
