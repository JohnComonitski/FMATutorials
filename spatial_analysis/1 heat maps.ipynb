**Import FMA Library**

```python
from FootballMatchAnalysis.objects.match import Match
from FootballMatchAnalysis.objects.plot import Plot
```

**Load Event and Tracking Data**

```python
DATADIR = './../data'
game_id = 1
```

```python
match = Match(DATADIR, game_id)
```

**Get a Player**

Players from either team can be retrieved using the Match object's `home_team` or `away_team` method.

```python
home_team = match.home_team()
player = home_team[5]
```

**Get a Player's Path**

The Match object has a `path` method, which takes a Player or Ball object and returns the path for that object throughout the match. The path will be returned as a list of coordinates. The `path` method can also take a `start_frame` and `end_frame` parameter if you just want the object's path over a certain period of time.

```python
# Path for Whole Match
path = match.path(player)
path
```

```python
# Path From Frame 1 to 2000
path = match.path(player, 1, 2000)
path
```

**Plot the Player's Path**

```python
plot = Plot()
plot.draw_path(path, c="b")
```

**Creating a Heat Map**

Plotting a player's path throughout a whole match will take a long time and likely be a pointless visualization. A more meaningful visualization is a heat map.

```python
# Get Path
path = match.path(player)

# Plot Path as Heat Map
plot = Plot()
plot.draw_heat_map(path)
```
