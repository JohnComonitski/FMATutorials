**Import FMA Library**

**Load Event and Tracking Data**

```python
DATADIR = './../data'
game_id = 1
```

```python
match = Match(DATADIR, game_id)
```

**Import the xT Library**

The FMA library includes a series of libraries that you can utilize to support your analysis. We will start by importing the `xt` library.

```python
from FootballMatchAnalysis.analysis.xt import *
```

**Visualize xT**

The Plot object has a `draw_xt` method, which will overlay onto the xT model used by the FMA library

```python
plot = Plot()
plot.draw_xt()
```

**Get xT of a Pass's Final Location**

```python
# Get a Pass
passes = match.get_events("PASS")
p = passes.iloc[11]

# Get Pass's Ending Location
start = (p["End X"], p["End Y"])

# Get xT of Passes Endling Location
xt = get_xt(start)
print(f"{p["From"]}'s pass left in the ball in a possition with an xt of {xt}")

plot.draw_event(p)
```

**Inverting the xT Plot**

You may notice that the xT plot progresses from left to right. This means it will work for the home team who is moving in that direction, but the away team will have their xT be misscalculated. The correct way for this is to include the `invert` parameter as `True` when calling `get_xt`.

```python
# Plot xT Inverted
plot = Plot()
plot.draw_xt(invert=True)

# Get a Pass
passes = match.get_events("PASS")
p = passes.iloc[30]

# Get Pass's Ending Location
start = (p["End X"], p["End Y"])

# Get xT of Passes Endling Location
xt = get_xt(start, invert=True)
print(f"{p["From"]}'s pass left in the ball in a possition with an xt of {xt}")

plot.draw_event(p, c='b')
```
