**Import FMA Library**

```python
from FootballMatchAnalysis.objects.match import Match
from FootballMatchAnalysis.objects.plot import Plot
```

**Load Event and Tracking Data**

```python
DATADIR = './../data'
game_id = 1
```

```python
match = Match(DATADIR, game_id)
```

**Import the xG Library**

The FMA library includes a series of libraries that you can utilize to support your analysis. We will start by importing the `xg` library.

```python
from FootballMatchAnalysis.analysis.xg import *
```

**Visualize xG**

The Plot object has a `draw_xg` method, which will overlay onto the xG model used by the FMA library

```python
plot = Plot()
plot.draw_xg()
```

**Get xG of Shots**

```python
# Get a Shot
shots = match.get_events("SHOT")
shot = shots.iloc[10]

# Get Shot's Starting Location
start = (shot["Start X"], shot["Start Y"])

# Get xG of Shot
xg = get_xg(start)
print(f"{shot["From"]}'s shot had an xG of {xg}")

plot.draw_event(shot)
```

**Inverting the xG Plot**

You may notice that the xG plot progresses from left to right. This means it will work for the home team who is moving in that direction, but the away team will have their shots xG's be misscalculated. The correct way for this is to include the `invert` parameter as `True` when calling `get_xg`.

```python
# Plot xG Inverted
plot = Plot()
plot.draw_xg(invert=True)

# Get a Shot
shots = match.get_events("SHOT")
shot = shots.iloc[2]

# Get Shot's Starting Location
start = (shot["Start X"], shot["Start Y"])

# Get xG of Shot
xg = get_xg(start, invert=True)
print(f"{shot["From"]}'s shot had an xG of {xg}")

plot.draw_event(shot, c='b')
```
